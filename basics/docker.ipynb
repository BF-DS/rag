# Introduction to Docker

Docker is a platform that allows you to **build, ship, and run applications** inside lightweight, portable containers. Containers package your application code together with its dependencies, ensuring it runs consistently across different environments.

In this notebook, we will cover:
1. What Docker is and why it's useful
2. Basic Docker concepts
3. Common Docker commands
4. Building and running a simple container

## 1. Why Docker?
- **Consistency**: Works the same in development, testing, and production.
- **Isolation**: Each container runs independently.
- **Lightweight**: Containers share the host OS kernel, making them faster and smaller than virtual machines.
- **Portability**: Run anywhere Docker is installed.


## 2. Docker Concepts
- **Image**: A template used to create containers. Think of it as a blueprint.
- **Container**: A running instance of an image.
- **Dockerfile**: A text file with instructions to build a Docker image.
- **Docker Hub**: A public registry for sharing Docker images.

## 4. Example: Creating a Simple Dockerfile
A **Dockerfile** is a text file with instructions to build a Docker image. 
Type the following command to create a new Dockerfile:
```bash
nano Dockerfile
```
This only works in bash terminal and not in Windows Command Prompt or PowerShell. Use `Ctrl + S` to save the file and `Ctrl + X` to exit the editor.



Here's an example of a simple Dockerfile:

```dockerfile
# Use Python base image
FROM python:3.9-slim

# Set working directory
WORKDIR /app

# Copy local files to container
COPY . .

# Install dependencies
RUN pip install --no-cache-dir -r requirements.txt

# Run the application
CMD ["python", "app.py"]
```

### Build and Run:
```bash
# Build the Docker image
docker build -t my-python-app .

# Run the container
docker run -p 5000:5000 my-python-app
```
